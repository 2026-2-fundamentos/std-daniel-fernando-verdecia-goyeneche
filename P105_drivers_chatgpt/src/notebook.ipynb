```python
import json
from pathlib import Path


def buscar_actividad():
    puntos_de_partida = []
    if "__file__" in globals():
        puntos_de_partida.append(Path(__file__).resolve().parent)
    puntos_de_partida.append(Path.cwd().resolve())

    for inicio in puntos_de_partida:
        for carpeta in [inicio, *inicio.parents]:
            if (carpeta / "notebooks").is_dir() and (carpeta / "tests").is_dir():
                return carpeta

    raise FileNotFoundError(
        "No encontré la carpeta de la actividad. Ejecuta el script desde "
        "dentro de la actividad (donde están las carpetas notebooks y tests)."
    )


def cuaderno_base():
    return {
        "cells": [],
        "metadata": {
            "kernelspec": {
                "display_name": "Python 3",
                "language": "python",
                "name": "python3",
            },
            "language_info": {"name": "python"},
        },
        "nbformat": 4,
        "nbformat_minor": 5,
    }


def celdas_de_participacion(con_id):
    celdas = [
        {
            "cell_type": "markdown",
            "metadata": {},
            "source": [
                "# Demostración guiada\n",
                "\n",
                "Participé siguiendo la demostración paso a paso junto con el profesor.",
            ],
        },
        {
            "cell_type": "code",
            "execution_count": None,
            "metadata": {},
            "outputs": [],
            "source": ['print("Participé en la demostración guiada")'],
        },
    ]
    if con_id:
        for numero, celda in enumerate(celdas, start=1):
            celda["id"] = f"participacion{numero}"
    return celdas


def cargar_cuaderno(ruta):
    if not ruta.is_file():
        return cuaderno_base()
    try:
        contenido = json.loads(ruta.read_text(encoding="utf-8"))
    except json.JSONDecodeError:
        return cuaderno_base()
    if not isinstance(contenido, dict):
        return cuaderno_base()
    if not isinstance(contenido.get("cells"), list):
        contenido["cells"] = []
    return contenido


def preparar_cuaderno(ruta):
    cuaderno = cargar_cuaderno(ruta)
    if len(cuaderno["cells"]) == 0:
        con_id = cuaderno.get("nbformat_minor", 0) >= 5
        cuaderno["cells"].extend(celdas_de_participacion(con_id))
    ruta.parent.mkdir(parents=True, exist_ok=True)
    ruta.write_text(
        json.dumps(cuaderno, indent=1, ensure_ascii=False) + "\n",
        encoding="utf-8",
    )


def main():
    actividad = buscar_actividad()
    rutas = [actividad / "notebooks" / "notebook.ipynb"]

    profesor = actividad / "professor" / "notebook.ipynb"
    if profesor.is_file():
        rutas.append(profesor)

    for ruta in rutas:
        preparar_cuaderno(ruta)
        print(f"Notebook listo: {ruta}")


main()
```
